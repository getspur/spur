### Full VO script (dry, one voice) — ~60s · ~145 words

**[0–4s · S01 open]**  
SPUR Context Service is cloud code context for third-party packages.  
*(C01)*

**[4–15s · S02 problem]**  
Inside your repo, every call site resolves. The moment a symbol crosses into a dependency, the edge goes dark — and the agent guesses. Without structured external context, agents fall back to web search, outdated docs, or guessing APIs across versions. Wrong assumptions break integrations, miss breaking changes, and force rework.  
*(C02, C06, C07)*

**[15–28s · S03 two planes]**  
Two planes, cleanly separated. In the worktree — this repository — you use knowledge context pack and code tools. For dependencies, you use the external MCP tools served by Context Service. Local tools answer what’s in this repo. External tools answer what’s in dependency X at version Y.  
*(C03, C04, C05)*

**[28–36s · S04 identity]**  
Every indexed unit is source, package, and revision. Selectors pin the version — for example, serde at one-point-oh-dot-one-nine-seven, trait Deserialize. No stale docs. No cross-version drift.  
*(C08)*  
*Pronunciation:* serde = “sir-dee” · `1.0.197` = “one point oh one nine seven” · `pkg:` = “package selector”*

**[36–50s · S05 tool loop]**  
Same graph-first workflow outside the worktree: orient with external knowledge context, then read real source, walk callers and callees. If a revision is cold, index on demand, keep working, and poll status — don’t block the session. Context complements local intelligence; it does not replace worktree code tools. Never feed package selectors into worktree reads.  
*(C09, C10, C11)*

**[50–60s · S06 CTA]**  
Version-precise answers without cloning every crate. Open context dot get spur dot dev.  
*(C11, C01)*
