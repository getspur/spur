# SPUR launch video: concept board

- **Surface:** Interactive HTML/JS concept selector inside Jute
- **Audience:** Product owner choosing the creative direction for a SPUR launch promo
- **Source:** `README.md`, treated as product truth
- **Brand context:** "The control tower for your CLI coding agents"
- **Tone:** Developer-native, precise, high-control, no generic AI theater
- **Scale:** Five genuinely different launch concepts, each with its own silhouette
- **Visual system:** Mission Control, deep navy, amber telemetry, cyan active state
- **Current lock:** Promotional launch video, not a product tour

The final video destination, duration, and run shape remain intentionally open until a concept is chosen.

## Concept-board plan

1. Present all five concepts as distinct visual worlds.
2. Make each concept navigable and animatable in one self-contained HTML artifact.
3. Keep the recommendation hidden until all five directions have been viewed.
4. Let the viewer select a direction locally, then confirm the number in chat.
5. Preserve a useful static view when active content is disabled.

## Refined direction: the TUI is the control tower

**Core message:** Work with one brain. See and steer every worker.

**Hierarchy**
1. The SPUR TUI owns the full frame and remains the primary interface.
2. The user works directly with the brain in the main conversation pane.
3. The brain routes execution through one of three visible modes: Plan, Loop, or Ad hoc.
4. Worker agents appear in an observation pane with lineage, state, cost, and review status.
5. Decisions and worker outcomes return to the same TUI for human approval.

**Design consequence:** The launch should never cut away to an abstract orchestration cloud. The product interface itself proves the control-tower claim.

## Theme comparison: human-centered SPUR

**Review logic now locked for the concept:** workers complete tasks, the brain reviews first, and only critical decisions escalate to the human.

**Comparison goal:** retain the credibility of a developer tool while removing cold mission-control styling, decorative telemetry, and generic AI aesthetics.

- **Warp Warm Dark:** terminal-native, calm, tactile, approachable.
- **Cursor Craft:** warm light canvas, editorial typography, product as software craft.

## Implementation-grounded UI lock

The current TUI establishes a more specific launch story than the earlier generic control-tower mock.

### Screen hierarchy

1. **Dashboard:** a vertical control surface with **Lineage**, then activity or a selected agent detail pane, then the composer and status bar.
2. **Focused worker:** the detail pane exposes the real tabs: `stream`, `artifacts`, `attempts`, `task`, and `review`.
3. **Brain session:** the primary working surface is **Dashboard > agent (brain)** with elapsed time and cost, a compact plan pulse, the ReAct trace, an optional live **Workers** band, the composer, and the status bar.
4. **Run shapes:** Plan, Loop, and Ad hoc change the work being observed. They should not be invented as permanent product tabs in these two screens.
5. **Review:** routine worker output is reviewed by the brain in the session trace. A critical conflict becomes the exceptional yellow human review card with `[A] approve [D] deny [M] modify+approve [R] retry`.

### Visual direction

**Warm terminal documentary.** Preserve the implementation's real dense, vertical terminal geometry and native cyan, green, yellow, and red semantics. Place it on a quiet warm-paper launch canvas with editorial annotations outside the product frame. No glowing nodes, fake radar, glass cards, or invented metrics.
